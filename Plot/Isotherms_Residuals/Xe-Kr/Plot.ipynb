```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#======= Read in the GCMC, TAL Prediction and IAST Prediction data ========
# P-X
df = pd.read_csv('../../../Datasets/Cu-BTC/Xe-Kr/Mixture_px.csv')
df2 = pd.read_csv('../../../IAST_Calculation/Cu-BTC/Xe-Kr/iast_pred_px.csv')
TAL_px = pd.read_csv('Pred_TAL_seed0.csv')

# P-X-T
df3 = pd.read_csv('../../../Datasets/Cu-BTC/Xe-Kr/Mixture_pxt.csv')
df4 = pd.read_csv('../../../IAST_Calculation/Cu-BTC/Xe-Kr/iast_pred_pxt.csv')
TAL_pxt = pd.read_csv('Pred_TAL_pxt_seed3.csv')

# # AL
# gf = pd.read_csv('Pred_AL_Xe.csv')
# gf2 = pd.read_csv('Pred_AL_pxt.csv')
```

```python
#======= Define Plot Parameters and Plot Data ========
mole_frac = [0.02, 0.06, 0.1, 0.2, 0.4, 0.6, 0.8, 0.9, 0.98]

# P-X
row_labels = ["(a)", "(b)", "(c)"]
row_y_positions = [0.917, 0.611, 0.302]

fig, axes = plt.subplots(figsize = (20, 18), nrows=3, ncols=3)

for ax, i in zip(axes.flat, mole_frac):
    # GCMC
    press_gcmc = df[df['Xe_mole_frac'] == i]['Pressure(Pa)'] / 1e5
    upt_xe_gcmc = df[df['Xe_mole_frac'] == i]['Xe_uptake']
    upt_kr_gcmc = df[df['Xe_mole_frac'] == i]['Kr_uptake']
    error_xe = df[df['Xe_mole_frac'] == i]['Xe_Error']
    error_xe = 2 * error_xe
    error_kr = df[df['Xe_mole_frac'] == i]['Kr_Error']
    error_kr = 2 * error_kr

    # IAST
    press_iast = df2[df2['Xe_mole_frac'] == i]['Pressure(Pa)'] / 1e5
    upt_xe_iast = df2[df2['Xe_mole_frac'] == i]['IAST_Xe_uptake']
    upt_kr_iast = df2[df2['Xe_mole_frac'] == i]['IAST_Kr_uptake']

    # TAL
    press_tal = TAL_px[TAL_px['Xe_mole_fraction'] == i]['X Test']
    upt_xe_tal = TAL_px[TAL_px['Xe_mole_fraction'] == i]['Xe_Pred']
    upt_kr_tal = TAL_px[TAL_px['Xe_mole_fraction'] == i]['Kr_Pred']


    # Xe
    ax.plot(press_gcmc, upt_xe_gcmc, marker='o', markersize='6', markeredgecolor='darkmagenta', linestyle=':', linewidth='4', color='purple', label="Xe (GCMC)")
    ax.fill_between(press_gcmc, (upt_xe_gcmc - error_xe), (upt_xe_gcmc + error_xe),color='gray', alpha=0.4)
    ax.plot(press_iast, upt_xe_iast, marker='s', markersize='6', markeredgecolor='mediumvioletred', linestyle=':', linewidth='4', color='pink', label="Xe (IAST)")
    ax.plot(press_tal, upt_xe_tal, marker='p', markersize='6', markeredgecolor='r', linestyle=':', linewidth='4', color='crimson', label="Xe (TAL)")
    # ax.plot(press_al, upt_xe_al, marker='p', markersize='6', markeredgecolor='r', linestyle=':', linewidth='4', color='yellow', label="Xe (AL)")
    
    # Kr
    ax.plot(press_gcmc, upt_kr_gcmc, marker='^', markersize='6', markeredgecolor='b', linestyle=':', linewidth='4', color='steelblue', label="Kr (GCMC)")
    ax.plot(press_iast, upt_kr_iast, marker='P', markersize='6', markeredgecolor='g', linestyle=':', linewidth='4', color='mediumseagreen', label="Kr (IAST)")
    ax.plot(press_tal, upt_kr_tal, marker='D', markersize='6', markeredgecolor='c', linestyle=':', linewidth='4', color='cyan', label="Kr (TAL)")
    # ax.plot(press_al, upt_kr_al, marker='D', markersize='6', markeredgecolor='r', linestyle=':', linewidth='4', color='orange', label="Kr (AL)")
    ax.fill_between(press_gcmc, (upt_kr_gcmc - error_kr), (upt_kr_gcmc + error_kr),color='gray', alpha=0.4,label="95% Confidence")
    
    ax.grid(True, linestyle='--', alpha=0.4)
    ax.set_title("Adsorption Isotherms Comparison at $X_{Xe}$ = " + f'{i}', fontsize = 17)
    ax.set_xlabel("Pressure (bar)", fontsize = 15, fontweight="bold")
    ax.set_ylabel("Adsorption Uptake (mg/g)", fontsize = 15, fontweight="bold")
    for tick in ax.get_xticklabels() + ax.get_yticklabels():
        tick.set_fontsize(13)
        tick.set_fontweight("bold")

handles, labels = axes.flat[0].get_legend_handles_labels()
fig.legend(handles, labels, loc="upper center", ncol=7, fontsize=17, frameon=False, bbox_to_anchor=(0.5, 0.96))
# Leave room for legend at top and row labels at left
fig.subplots_adjust(left=0.075,right=0.98,bottom=0.055,top=0.88,wspace=0.25,hspace=0.35)
for label, y in zip(row_labels, row_y_positions):
    fig.text(0.025,y,label,fontsize=15,fontweight="bold",ha="left",va="center")
plt.tight_layout(rect=[0, 0, 1, 0.93])
plt.savefig('Adsorption_Isotherm_comp_px.png', dpi=600, bbox_inches="tight", pad_inches=0.08, facecolor="white")
plt.savefig('Adsorption_Isotherm_comp_px.pdf', dpi=600, bbox_inches="tight", pad_inches=0.08, facecolor="white")
plt.show()
```
